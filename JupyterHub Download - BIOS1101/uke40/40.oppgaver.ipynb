**a)**
Skriv en Python-funksjon som sjekker om en streng
er en kombinasjon av bare fire baser A, C, G og T
og som skriver ut en melding om at strengen er en DNA sekvens,
eller at strengen ikke er det.
Test funksjonen din på sekvensen nedenfor 
så vel som på en versjon av sekvensen som faktisk er en DNA sekvens.

````python
dna_sekvens = "ACTGGAGTCAGGBCCGT"

def er_dna(string):
    baser_ikkedna = 0
    for base in string:
        if base not in "ATCG":
            baser_ikkedna += 1
    if baser_ikkedna > 0:
        print("Dette er ikke DNA")
        return False
    else:
        print("Dette er DNA")
        return True

print(f"{er_dna(dna_sekvens)}")
````

*Valgfritt*: Bruk `er_DNA()` som navn på funksjonen og la return-verdien
være *enten* den boolske verdien `True` 
hvis strengen er en kombinasjon av bare fire baser A, T, C og G,
*eller* den boolske verdien `False`.


**b)**
Prøv å endre DNA-sekvensen i `dna_sekvens` og se om programmet ditt fungerer.

````python
dna_sekvens = "ATTCTGCGCCATTAG"
print(f"{er_dna(dna_sekvens)}")
````
