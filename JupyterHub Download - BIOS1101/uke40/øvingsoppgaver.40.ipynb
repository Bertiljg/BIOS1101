PCR can be used to amplify small amounts of DNA to much larger amounts, sufficient to do further molecular biological experiments. PCR is usually performed for a number of cycles. In each cycle, a new copy of each target DNA is made. If we assume a PCR is 100% efficient it *doubles* the number of target molecules every cycle. However, this is normally not the case. There are many factors, such as the total amount of reagents, that limit the efficiency of the reaction during different steps. 
A more realistic efficiency for a PCR reaction is 60%, meaning that, for each cycle, *60% of the original DNA gets copied and 20% does not*

For this exercise, you'll need these formulas:

* The formula for calculating the number of DNA molecules given a certain amount of DNA is

> Number of DNA molecules = (weight of DNA in gram \* Avogadro's number) / (length of DNA in bases \* 650 gram/mole)

* The formula for converting the number of DNA molecules of DNA into their weight is

> weight of DNA in gram = (number of DNA molecules \* length of DNA in bases * 650 gram/mole) / Avogadro's number

with Avogadro's number = $6.022\times10^{23}$

For the PCR reaction:
* 1 picogram, or $1\times10^{-12}$ gram, of the DNA to be amplified will be present at the beginning of the reaction
* the length of DNA to be amplified, in bases, is 1666
* the PCR will be performed using 30 cycles

Write a Python program that:
* calculates the number of copies of the target after the 30 cycles have ended, assuming 80% PCR efficiency, using a for loop
* calculates the final weight of amplified target DNA in nanograms, $1\times10^{-9}$ gram

## Oppgave 32: 'greedy' translation

Below is the `translate()` function developed in chapter 7 in the course book:

````
def translate(mRNA):
    """
    Translate a mRNA string to a protein.
    """

    codon_dict = {
        "UUU": "F", "UUC": "F", "UUA": "L", "UUG": "L",
        "UCU": "S", "UCC": "S", "UCA": "S", "UCG": "S",
        "UAU": "Y", "UAC": "Y", "UAA": "*", "UAG": "*",
        "UGU": "C", "UGC": "C", "UGA": "*", "UGG": "W",
        "CUU": "L", "CUC": "L", "CUA": "L", "CUG": "L",
        "CCU": "P", "CCC": "P", "CCA": "P", "CCG": "P",
        "CAU": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
        "CGU": "R", "CGC": "R", "CGA": "R", "CGG": "R",
        "AUU": "I", "AUC": "I", "AUA": "I", "AUG": "M",
        "ACU": "T", "ACC": "T", "ACA": "T", "ACG": "T",
        "AAU": "N", "AAC": "N", "AAA": "K", "AAG": "K",
        "AGU": "S", "AGC": "S", "AGA": "R", "AGG": "R",
        "GUU": "V", "GUC": "V", "GUA": "V", "GUG": "V",
        "GCU": "A", "GCC": "A", "GCA": "A", "GCG": "A",
        "GAU": "D", "GAC": "D", "GAA": "E", "GAG": "E",
        "GGU": "G", "GGC": "G", "GGA": "G", "GGG": "G"
    }

    index = mRNA.index("AUG")
    protein = ""
    amino_acid = ""
    while amino_acid != "*" and index < len(mRNA):
        # Convert a codon to an amino acid
        # and add it to our protein
        codon = mRNA[index:index + 3]
        # The last slice of the string may be shorter
        # if the length of the sequence to translate is not a multiple of 3 bases.
        # We thus need to check that the codon is 3 nucleotides.
        if len(codon) == 3:
            amino_acid = codon_dict[codon]
            protein += amino_acid
        index += 3

    return protein
````

Here is an example on how to use this function to translate a DNA sequence to protein:

````
dna = "CAATGGCAACATTTCAAGTCTTCCAATAAATAGGAC"
rna = dna.replace("T", "U")
protein = translate(rna)
print(protein)
````
