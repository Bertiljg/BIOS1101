Considering the following statements:

1. temperature in degrees Fahrenheit is the temperature in degrees Celsius times 9/5 plus 32
2. to calculate the temperature in degrees Kelvin, first subtract 32 from the temperature in degrees Fahrenheit, multiply the result with 5/9 and finally add 273.15
3. temperature in degrees Celsius is the temperature in degrees Kelvin minus 273.15
4. to calculate the temperature in degrees Celsius, subtract 32 from the temperature in degrees Fahrenheit and multiply the result with 5/9

Reorder the Python parts below so that they form four functions, each calculating one of the conversions listed above.
You don't need to add any other pieces of code,
except for the equal sign `=` where it is needed.
Each function consist of three lines of code.

````python
def celsius_to_fahrenheit(temp_c):
    fahrenheit = temp_c * 9/5 + 32
    return fahrenheit
def fahrenheit_to_celsius(temp_f):
    celsius = (temp_f - 32) * (5/9)
    return celsius
def fahrenheit_to_kelvin(temp_f):
    kelvin = ((temp_f - 32) * (5/9)) + 273.15
    return kelvin
def kelvin_to_celsius(temp_k):
    celsius = temp_k - 273.15
    return celsius

print(celsius_to_fahrenheit(37.78))
print(fahrenheit_to_kelvin(-279.67))
print(kelvin_to_celsius(373.15))
print(fahrenheit_to_celsius(212))
````

Lag et Python-program som

* avgjør med en if-else-setning om hver unge ble hann eller hunn
* teller antall hanner og antall hunner
* skriver ut de to antallene på skjermen


````python
Mann = []
Kvinne = []
kjønn = ["Mann", "Kvinne"]

N = 1000
n = 1

while n < N+1:
    avkom_kjønn = np.random.choice(kjønn)
    if avkom_kjønn == kjønn[0]:
        Mann.append(kjønn)
    else:
        Kvinne.append(kjønn)
    n += 1

if len(Mann) + len(Kvinne) == 1000:
    print(f"Hanner: {len(Mann)}, Hunner: {len(Kvinne)}")
else:
    print(f"noe gikk galt!")
````

## Oppgave 5: Overcoming strings being immutable

<div class="alert alert-block alert-info">
The goal of this exercise is to practice string slicing and become aware of what it means when a string is called 'immutable'.
</div>

Slicing a string works the same as slicing a list, for example:

````python
my_list = [100, 12, -3, 4.5, 62, 1.23]
print(my_list[3:5])
````

````python
my_string = 'Hello world!'
print(my_string[3:5])
````

However, there is one big difference between operating with list slices and string slices. With lists, you can do this:

````python
my_list = [100, 12, -3, 4.5, 62, 1.23]
my_list[4] = 501
print(my_list)
````

```Python
        [100, 12, -3, 4.5, 501, 1.23]
```

But you can not change a string 'in place':

````python
my_string = 'Hella world!'
my_string[4] = 'o'
print(my_string)
````

Strings are so-called 'immutable', they cannot be changed once they are created.
