You suspect that someone has meddeled with the coin so that it is biased to "tails" up. You know that a fair coin should have a 50% chance of landing on either side and would like to check this. You should toss the coin at least 1000 times to be able to tell whether it is fair or not. Make a program that simulates these coin tosses and that prints out the percentages of heads and tails (you could try using a for-loop and the range() function here). Does this seem like a fair coin? What happens if you lower the number of tosses? Or increase them? Discuss your results.

<!-- --- begin answer of exercise --- -->
**Fasit.**
When you throw the coin only a few times, unequal distribution is more likely just by chance. As you increase the number of tosses, the distribution will even out (the law of large numbers).
<!-- --- end answer of exercise --- -->

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

````
import numpy as np

coin = ["heads", "tails"]

heads_count = 0
tails_count = 0

for i in range(1000):
    toss_result = np.random.choice(coin)

    if toss_result == "heads":
        heads_count += 1
    if toss_result == "tails":
        tails_count += 1

heads_percentage = (heads_count/(heads_count + tails_count)) * 100
tails_percentage = (tails_count/(heads_count + tails_count)) * 100

print(f"The percentage of heads is {heads_percentage} and the percentage of tails is {tails_percentage}")
````

**d)**
Make a function that tosses a coin a desired numer of times
(that is, number of tosses is the argument)
and returns the percentages of heads and tails.
Round the percentage off to 2 decimals.
Call the function with, 10, 100, 1000 and 10000 tosses.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

````
def coin_toss(tosses):
    coin = ["heads", "tails"]

    heads_count = 0
    tails_count = 0

    for i in range(tosses):
        toss_result = np.random.choice(coin)

        if toss_result == "heads":
            heads_count += 1
        if toss_result == "tails":
            tails_count += 1

    heads_percentage = (heads_count/(heads_count + tails_count)) * 100
    tails_percentage = (tails_count/(heads_count + tails_count)) * 100

    return f"The percentage of heads is {heads_percentage :.2f} and the percentage of tails is {tails_percentage :.2f}"
````

````
print(coin_toss(10))
print(coin_toss(100))
print(coin_toss(1000))
print(coin_toss(10000))
````

The 'algorithm' is to, while you loop over the list, keep track of the 'maximum number you have seen so far' and check whether each next number in the list is higher than the 'maximum number you have seen so far'.
````
current_max = 0
for number in numbers:
    if number > current_max:
        current_max = number
print(current_max)
````

Here would be the function:

````
def my_max(my_numbers):
  current_max = 0
  for number in my_numbers:
      if number > current_max:
          current_max = number
  return current_max
````

## Oppgave 23: Factorials

The Factorial of a positive number *n*, represented as $n!$, is the product of whole numbers from 1 to n. i.e,

$n! = 1  \times 2 \times 3  \times  ..... \times  (n-1)  \times n$

For example, the factorial of 5 is $5!$:

$ 5! = 5 \times 4 \times 3 \times 2 \times 1 = 120$

## Oppgave 24: Fibonacci

Hva skrives ut når dette programmet kjøres?

````
import numpy as np

def fibonacci(N):
    x = np.zeros(N+1)
    x[0] = 1
    x[1] = 1
    for n in range(2, N+1):
        x[n] = x[n-1] + x[n-2]
    return x

print(max(fibonacci(5)))
````
