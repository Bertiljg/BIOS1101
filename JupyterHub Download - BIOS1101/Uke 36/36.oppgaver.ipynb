<details style="background-color: #eafaf1; padding: 10px; border-radius: 5px; border-left: 4px solid #27ae60;">
<summary style="cursor: pointer; font-weight: bold; color: #27ae60;">🤔 Komponentene til en While-Løkke</summary>

En fungerende `while`-løkke består av fire deler:

* **Initiering:** Definer løkkevariabelen før løkken starter, f.eks. `i = 0`
* **Betingelse:** Løkken kjører så lenge betingelsen er sann, f.eks. `while i < 10:`
* **Repetisjon:** Koden som skal kjøres hver gang, f.eks. `print(i)`
* **Oppdatering:** Endre løkkevariabelen slik at løkken til slutt stopper, f.eks. `i = i + 1`

</details>
