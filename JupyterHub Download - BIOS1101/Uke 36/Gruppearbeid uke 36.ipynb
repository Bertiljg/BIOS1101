```python
lengder = []
indeks = 0

while indeks < 4:
    print("Når indeks er", indeks, "er betingelsen", indeks < 4)
    sekvens = dna_sekvenser[indeks]
    lengde = len(sekvens)
    lengder.append(lengde)
    indeks = indeks + 1

print("Løkken er ferdig")
print("Når indeks er", indeks, "er betingelsen", indeks < 4)
print(lengder)
```
