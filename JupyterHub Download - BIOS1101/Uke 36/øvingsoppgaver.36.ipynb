## Oppgave 11: Viruses

Here we want to model biological exponential growth as in the case of DNA viruses. As part of a viral infection, viral DNA is replicated. DNA viruses replicate their DNA inside the cell they have infected using the DNA replication machinery of that cell during several replication cycles. Once enough copies of the virus DNA have been created, each DNA molecule is packaged into a new virus (simply put, inside a protein layer) and the host cell bursts, releasing all the viruses that then can infect new cells.

If we assume that the cellular DNA synthetic machinery is 100% efficient, the number of virus DNA molecules *doubles* every replication cycle. However, this is normally not the case. There are many factors that limit the efficiency of replication. A more realistic efficiency for viral DNA replication is 20%, meaning that, for each replication cycle, *20% of the original viral DNA gets copied and 80% does not*

For this exercise, you'll need these formulas:

* The formula for calculating the number of viral DNA molecules given a certain number of infected host cells is

> Number of viral DNA molecules = Number of viral DNA molecules per cell * number of infected cells

* replication efficiency is 20%, as explained above
* we will model 30 replication cycles before the cell bursts and releases the viruses

Write a Python program that:
* calculates the number of viral DNA molecules present in each infected cell after the 30 replication cycles have ended, assuming 20% replication efficiency
* calculate the total number of viruses that will be released in the population of cells if 102 cells are infected
