Print the variables `count` and `time_unit`.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

````python
count = 100
growth_rate = 2.3 #per time unit

time_unit = 0
while count < 3000: #one loop per time unit
  count = count * growth_rate
  time_unit = time_unit + 1

print("Sum:", count, "Time units:", time_unit)
````

## Oppgave 11: Viruses

Here we want to model biological exponential growth as in the case of DNA viruses. As part of a viral infection, viral DNA is replicated. DNA viruses replicate their DNA inside the cell they have infected using the DNA replication machinery of that cell during several replication cycles. Once enough copies of the virus DNA have been created, each DNA molecule is packaged into a new virus (simply put, inside a protein layer) and the host cell bursts, releasing all the viruses that then can infect new cells.

If we assume that the cellular DNA synthetic machinery is 100% efficient, the number of virus DNA molecules *doubles* every replication cycle. However, this is normally not the case. There are many factors that limit the efficiency of replication. A more realistic efficiency for viral DNA replication is 20%, meaning that, for each replication cycle, *20% of the original viral DNA gets copied and 80% does not*

For this exercise, you'll need these formulas:

* The formula for calculating the number of viral DNA molecules given a certain number of infected host cells is

> Number of viral DNA molecules = Number of viral DNA molecules per cell * number of infected cells

For the viral replication:
* 102 cells are infected
* each cells is infected by a single virus particle containing one virus DNA molecule at the beginning
* replication efficiency is 20%, as explained above
* we will model 30 replication cycles before the cell bursts and releases the viruses

Write a Python program that:
* calculates the number of viral DNA molecules present in each infected cell after the 30 replication cycles have ended, assuming 20% replication efficiency
* calculate the total number of viruses that will be released in the population of cells if 102 cells are infected

<!-- --- begin answer of exercise --- -->
**Fasit.**

Calculate number of viruses that will be produced in a single host over 30 replications

Using a loop:

````python
import numpy as np
virus = 1
total_viruses = np.zeros(31)
total_viruses[0] = virus

cycle = 1
while cycle < 31:
    total_viruses[cycle] = total_viruses[cycle - 1] + total_viruses[cycle - 1] * 0.2
    cycle = cycle + 1
final_viruses = total_viruses[-1] * 102
print(final_viruses)
````

Using direct calculations: final number of copies is $virus\times1.2^{30}$

````python
final_viruses = virus * 1.2**30 * 102
viruses_per_cell = virus * 1.2**30
print(final_viruses)
print(viruses_per_cell)
````

## Oppgave 12: Vekst i en bladluspopulasjon

I denne oppgaven skal du undersøke en biologisk vekstmodell,
bruke modellen i Python og vurdere antagelsene den bygger på.

En biolog undersøker veksten av en bladluspopulajson på planter i et drivhus.
Ved starten av forsøket er det **100 bladlus**.
Populasjonen undersøkes en gang per uke.
Bladluspopulajsonen vekst kan beskrives av følgende vekstmodell:

$$
N_t = s_aN_{t-1} + s_neN_{t-1}
$$

der:

%s
  :   
  $N_t$ er antall bladlus i uke $t$
%s
  :   
  $e=4$ er gjennomsnittlig antall nye bladlus per individ
%s
  :   
  $s_n=0.25$ er overlevelsesraten til de nyfødte
%s
  :   
  $s_a=0.60$ er overlevelsesraten til bladlusene som allerede finnes
%s
  :   
  $N_0=100$ er startverdien

**a)**
Hva slags vekstmodell beskriver populajsonsveksten til bladlusen? og Hva representerer de to leddene i vekstmodellen?

$$
s_aN_{t-1}
\qquad\text{og}\qquad
s_neN_{t-1}
$$

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**
Modellen er en eksponensiell vekstmodell. 
Leddet $s_aN_{t-1}$ representerer antall eksisterende bladlus fra uke $t-1$
som overlever til uke $t$.
Leddet $s_n e N_{t-1}$ Representerer antall nyfødte bladlus som overlever til uke $t$.
<!-- --- end solution of exercise --- -->

**b)**
Gitt initial populasjonen, beregn den predikerte bladuspopulajsonen en uke etter.

````python
# Modellparameterne
N0 = 100
e = 4
s_n = 0.25
s_a = 0.60
````

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

````python
# Modellparameterne
N0 = 100
e = 4
s_n = 0.25
s_a = 0.60

N1 = s_a * N0 + s_n * e * N0
print("Etter en uke er det" , N1, "bladlus.")
````

<!-- --- end solution of exercise --- -->

**c)**
Forskeren er interessert i hvordan modellen predikerer populajsonstanden etter 10 uker med vekst.

Bruk en løkke for å simulere populasjonstørrelsen gjennom 10 uker. Hvor stor er bladluspopulajsonen etter 10 uker?

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

````python
# Modellparameterne
N0 = 100
e = 4
s_n = 0.25
s_a = 0.60

uke = 1
N = N0

# Beregner populasjonen uke for uke
while uke <= 10:
    ny_N = s_a * N + s_n * e * N
    N = ny_N
    uke = uke + 1

print("Etter 10 uker er det", N, "bladlus")
````

<!-- --- end solution of exercise --- -->

**d)**
Visualiser bladlusveksten over de 10 første ukene i et python plot.

**Plottet må inkludere:**

* beskrivende tittel
* navn på aksene

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

````python
import matplotlib.pyplot as plt

# Modellparameterne
N0 = 100
e = 4
s_n = 0.25
s_a = 0.60
antall_uker = 10

# Lister for plotting
uker = []
uker.append(0)
N = [] # N er nå en liste
N.append(N0)

uke = 1
while uke <= 10:
    ny_N = s_a * N[uke - 1] + s_n * e * N[uke -1]
    
    # Legger til de nye verdiene i listene
    uker.append(uke)
    N.append(ny_N)

    uke = uke + 1

# Plotting
plt.plot(uker, N)
plt.title('Vekst av bladluspopulasjon over 10 uker')
plt.xlabel('Uke')
plt.ylabel('Antall bladlus')
plt.show()
````
