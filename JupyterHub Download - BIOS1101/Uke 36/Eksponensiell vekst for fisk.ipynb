```python
import os as os
#definere listene som skal brukes
uke36_csv = []
uke36 = os.listdir()

#angir variabler for hva som skal søkes etter og foreslår feilhjelp
try:
    position = input("skal filtypen starte eller slutte med et søkeord? Skriv start eller slutt")
    if position == "start":
        funksjon = str.startswith
    elif position == "slutt":
        funksjon = str.endswith

    filtype = input("skriv prefix streng om du leter etter prefix, skriv suffix filtype om man leter etter suffix")
except:
    print("man kan kun skrive start eller slutt.")

#løkkevariabler
a = len(uke36)
i = 0

print("lengden på listen av alle mappeobjekter er:", a)

#løkke for å lage en ny liste
while i < a:
    streng = funksjon(uke36[i], filtype)
    if streng == True:
        uke36_csv.append(uke36[i])
        print(uke36[i], "er en fil som starter/slutter med", filtype)
        print(streng)
    else:
        print(uke36[i], "ender ikke på", filtype)

    i = i + 1
print("av de", len(uke36), "elementene i mappen", os.getcwd(), ", hadde ", len(uke36_csv), "endingen", filtype, "eller starten", position)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


#Løkkevariabler
i = 0
a = len(uke36_csv)
listeliste = []

#Løkken går fra start til slutt i listen uke36_csv og lager nye lister i variablene "t" og "E". Disse listene blir så lagret i en ny liste med navn "listeliste"
while i < a:
    data = pd.read_csv(uke36_csv[i])
    t = list(data["t"])
    E = list(data["E"])
    listeliste.append(t)
    listeliste.append(E)
    i = i + 1

#nye løkkevariabler
i_2 = 0
b = len(listeliste)

#løkken går fra start til slutt i lista "listeliste" og lagrer plot av hvert eksperiment. Siden i_2 må økes med 2 for hvert steg deler vi på 2 for å finne rett etikett i listen uke36_csv.
while i_2 < b:
    plt.plot(listeliste[i_2], listeliste[i_2+1], "o-", label=(uke36_csv[i_2 // 2]))
    i_2 = i_2 + 2

#plottefunksjoner
plt.yscale("log")
plt.xlabel("Tid, t (minuter)")
plt.ylabel("Populasjonsstørrelse, E")
plt.title("Bakteriell vekst")
plt.legend()
plt.show()
```
