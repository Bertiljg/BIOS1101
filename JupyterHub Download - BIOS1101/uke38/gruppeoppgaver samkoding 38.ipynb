```python
# %load code/maur.py
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("ant_species_elevation.csv")

høyde_over_havet = list(data["Elevation"])
maurarter = list(data["Number of ant species"])

plt.plot(høyde_over_havet, maurarter, "o")
plt.title("Antallet maurarter ved ulike høydemeter i Spring Mountains")
plt.xlabel("Meter over havet (m)")
plt.ylabel("Antallet maurarter")
plt.show()
```

```python
høyde_beregnet =[1300, høyde_over_havet[-1]]
```

```python
maur_beregnet = []

maur_beregnet.append(1300 * 0.02 - 16)
maur_beregnet.append(høyde_over_havet[-1] * 0.02 - 16) #høyde 2000 beregnet maurart mengde
```

```python
print(maur_beregnet)
```

```python
plt.plot(høyde_over_havet, maurarter, "o")
plt.plot(høyde_beregnet, maur_beregnet, label = "lineær modell")
plt.title("Antall maurarter i høydemeter over havet (Spring Mountain)")
plt.xlabel("Meter over havet (m)")
plt.ylabel("Antall maurarter")
plt.legend()
plt.show()
```

```python
import numpy as np
a, b = np.polyfit(høyde_over_havet[3:11], maurarter[3:11], 1)
print("Stigningstall:", round((a), 3), "Skjæringspunkt:", round((b), 3))
```

```python
høyde_fit = np.linspace(1000, 2200, 100)
print(høyde_fit)
```

```python
type(høyde_fit)
```

```python
maur_fit = a * høyde_fit + b
```

```python
print(maur_fit)
```

```python
plt.plot(høyde_over_havet, maurarter, "o")
plt.plot(høyde_fit, maur_fit, ":", label = "lineær modell")
plt.title("Antall maurarter i høydemeter over havet (Spring Mountain)")
plt.xlabel("Meter over havet (m)")
plt.ylabel("Antall maurarter")
plt.legend()
plt.show()
```

```python
a2, b2, c2 = np.polyfit(høyde_over_havet, maurarter, 2)
maur_fit2 = a2 * høyde_fit**2 + b2*høyde_fit + c2
plt.plot(høyde_over_havet, maurarter, "o")
plt.plot(høyde_fit, maur_fit, ":", label="lineær regresjonsmodell")
plt.plot(høyde_fit, maur_fit2, "-", label="Kvadratisk regresjonsmodell")
plt.title("Antall maurarter i høydemeter over havet (Spring Mountain)")
plt.xlabel("Meter over havet (m)")
plt.ylabel("Antall maurarter")
plt.legend()
plt.show()
```

```python
a3, b3, c3, d3 = np.polyfit(høyde_over_havet, maurarter, 3)
maur_fit3 = a3 * høyde_fit**3 + b3 * høyde_fit**2 + c3 * høyde_fit + d3

plt.plot(høyde_over_havet, maurarter, "o")
plt.plot(høyde_fit, maur_fit, ":", label="lineær regresjonsmodell")
plt.plot(høyde_fit, maur_fit2, "-", label="Kvadratisk regresjonsmodell")
plt.plot(høyde_fit, maur_fit3, "--", label = "Kubist regresjonsmodell")

plt.title("Antall maurarter i høydemeter over havet (Spring Mountain)")
plt.xlabel("Meter over havet (m)")
plt.ylabel("Antall maurarter")
plt.legend()
plt.show()
```
