````
class Student:
    def __init__(self, navn, alder, kurs):
        self.navn = navn
        self.alder = alder
        self.kurs = kurs
        self.bestått = {}
        for individuelle_kurs in kurs:
            self.bestått[individuelle_kurs] = False  
        self.snakker_med = []  # Liste for å holde pekere til andre Student-objekter

    def øk_alder(self):
        self.alder += 1

    def sjekk_alder(self):
        return self.alder 

    def sjekk_bestått(self, kursnavn):
        if kursnavn in self.bestått:
            return self.bestått[kursnavn]                        
        else:
            return f"Kurs '{kursnavn}' ikke funnet."

    def sett_bestått(self, kursnavn):
        if kursnavn in self.bestått:
            self.bestått[kursnavn] = True
        else:
            print(f"Kurs '{kursnavn}' ikke funnet.")

    def legg_til_snakker_med(self, andre_student):
        if andre_student not in self.snakker_med:
            self.snakker_med.append(andre_student)
            andre_student.snakker_med.append(self)

    def avslutt_samtale_med(self, andre_student):
        # Fjern den doble forbindelsen
        if andre_student in self.snakker_med:
            self.snakker_med.remove(andre_student)
            andre_student.snakker_med.remove(self)

    def __str__(self):
        bestått_status = ""
        for kurs, status in self.bestått.items():
            if status:
                svar='Ja'
            else:
                svar='Nei'
            bestått_status += f"  {kurs}: {svar}\n"

        snakker_med_navn = ""
        for student in self.snakker_med:
            snakker_med_navn += student.navn + ", "

        if snakker_med_navn:
            snakker_med_navn = snakker_med_navn[:-2]  # Fjern siste komma og mellomrom
        
        snakker_med_display = snakker_med_navn if snakker_med_navn else 'Ingen'

        beskjed = f"Navn: {self.navn}\n"
        beskjed += f"Alder: {self.alder}\n"
        beskjed += f"Kurs bestått:\n{bestått_status}"
        beskjed += f"Snakker med: {snakker_med_display}"

        return beskjed
````

````
class Hund:
    def __init__(self, rase, farge, alder):
        self.rase = rase
        self.farge = farge
        self.alder = alder

    def bjeff(self):
        print("Voff!")
````

````
coco = Hund("dvergpuddel", "grå", 4)
tarzan = Hund("bulldog", "hvit", 6)
````

Hundene snakker!

````
coco.bjeff()
tarzan.bjeff()
````

Viktig å legge merke til:

* En klassedefinisjon starter med nøkkelordet `Class` og klassens navn
* Alle klasser skal ha en funksjon `__init__`
* Denne funksjonen har to understrekingstegn før og etter navnet, dette indikerer at det er en spesialfunksjon for klasser
* Klassen defenieres (vanligvis) med en rekke funksjoner
* Variabelen `self` dukker opp i begge funksjoner. Denne representerer instansen av klassen,  altså 'dette objektet'. `self` må være første parameter for alle funksjoner i en klasse,
   ellers fungerer ikke objektene som de skal.
* `__init__` kalles *konstruktøren*: funksjonen som initierer instansen (objektet) og brukes vanligvis til å sette en rekke variabler som instansen skal ha
* disse variablene heter instansvariabler, og representerer egenskapene, eller attributtene til instansen (objektet)

Merk at siden `alder` er en instans variabel skal vi ikke ha parentes `()`
bak variabelnavnet.

Men å bruke instansvariabler på denne måten er ikke god praksis.
God praksis er å lage en funksjon for dette i klassens definisjon:

````
class Hund:
    def __init__(self, rase, farge, alder):
        self.rase = rase
        self.farge = farge
        self.alder = alder

    def bjeff(self):
        print("Voff!")

    def hent_alder(self):
        return self.alder
````

````
class Hund:
    def __init__(self, rase, farge, alder):
        self.rase = rase
        self.farge = farge
        self.alder = alder

    def bjeff(self):
        print("Voff!")

    def hent_alder(self):
        return self.alder

    def hent_rase(self):
        return self.rase

tarzan = Hund("bulldog", "hvit", 6)
print(f"Tarzan er {tarzan.hent_alder()} år og av rasen {tarzan.hent_rase()}.")
````

Tarzan har vært hos oss en stund nå og har dermed blitt et år eldre.
Hvordan kan vi oppdatere instansen?

Til det trenger vi en ny metode.

````
class Hund:
    def __init__(self, rase, farge, alder):
        self.rase = rase
        self.farge = farge
        self.alder = alder

    def bjeff(self):
        print("Voff!")

    def hent_alder(self):
        return self.alder

    def hent_rase(self):
        return self.rase

    def øk_alder(self):
        self.alder += 1
````

````
tarzan = Hund("bulldog", "hvit", 6)
print(f"Tarzan er {tarzan.hent_alder()} år.")
# tarzan har bursdag!
tarzan.øk_alder()
print(f"Tarzan er {tarzan.hent_alder()} år.")
````

Legg merke til

* funksjonen `øk_alder` har *ikke* en return verdi!
