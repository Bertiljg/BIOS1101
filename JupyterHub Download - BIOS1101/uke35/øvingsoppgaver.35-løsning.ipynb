## Oppgave 4: Simple statistics

**a)**
A group of students reported basic physical measurements. Use the following lists to store the information given:

```python
height = [176, 169, 172, 172, 179, 160, 157, 170, 190, 154, 181, 185, 168, 184]
age = [34, 33, 27, 31, 19, 34, 22, 28, 23, 32, 20, 20, 24, 30, 21, 25]
weight = [62, 46, 56, 52, 80, 51, 84, 65, 92, 55, 53, 52, 84, 84, 61]
```

**b)**
Quite often, data you receive from others will have problems, so you need to do some quality control fist.

Write a program that prints the length (number of items) of all the lists. If they do NOT all consist of the same amount of items, use the approriate method to delete the last element(s) of the lists that are too long. Then print their new lengths again to verify.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```python
print(len(height))
print(len(age))
print(len(weight))
del age[-2:]
del weight[-1]

print(len(height))
print(len(age))
print(len(weight))
```

<!-- --- end solution of exercise --- -->

**c)**
Write the sums of all the values in each list to separate variables.

**Hint.**

```python
heightsum = sum(height)
agesum = sum(age)
weightsum = sum(weight)
```

<!-- --- end solution of exercise --- -->

**d)**

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**
Different solutions are possible, rounding can also be done within the `str` command, for example. Here is one solution:

## Oppgave 8: Gibberellin and plant growth

Gibberellins are plant hormones involved in processes such as germination and regulation of cell elongation. Exposing plants to gibberellins causes them to grow taller, and plants have been successfully bred to produce less gibberellin in order to grow shorter. In this exercise you will handling *real* data from a simple experiment with gibberellin in pea plants and see whether you can estimate the amount of gibberellin in a solution with an unknown concentration.

**a)**
The file *gibberellin.csv* contains the concentration of gibberellin (in molar) applied to dwarf pea plants, and their subsequent growth (in cm) over a period of two weeks. Import the file, inspect the rersults. Then, convert the data from the two columns into separate lists with appropriate names.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**
Inspecting the data:

```python
import pandas as pd

plants_growth = pd.read_csv("gibberellin.csv")
plants_growth
```

Extracing the lists:

```python
concentration = list(plants_growth["gibberellin"])
growth = list(plants_growth["growth"])
print(concentration)
print(growth)
```

<!-- --- end solution of exercise --- -->

**b)**
Plot the growth data against the gibberellin data and take note of what trend is observed. Do you think it is suitable to use a logarithmic axis, and if so, which one?

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```python
import matplotlib.pyplot as plt

plt.plot(concentration, growth, label="Lengthwise growth")
plt.xlabel("Concentration of gibberellin")
plt.ylabel("Growth in centimeters")
plt.title("Lengthwise growth as a function of gibberellin concentration")
plt.show()
```

The gibberellin concentration is exponential with base 10, so it would make sense to plot the x-axis on a log scale. That still does not make the relationship linear, though.

```python
plt.plot(concentration, growth, label="Lengthwise growth")
plt.xlabel("Concentration of gibberellin")
plt.ylabel("Growth in centimeters")
plt.xscale("log")
plt.title("Lengthwise growth as a function of gibberellin concentration")
plt.show()
```
