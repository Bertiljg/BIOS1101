<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```
import pandas as pd

# Spørsmål 1: gjennomsnittlig temperatur over tid rundet ned til to desimaler
# Leser av fra csv og konverteter dataramme-objektet om til en liste
data = pd.read_csv("temperatures.csv")
temperatur_data = list(data["Temperature"])

snitt_temp = sum(temperatur_data)/len(temperatur_data)
print(round(snitt_temp,2))

# Spørsmål 2: tidsintervallet brukt i temperaturmålingene oppgitt i minutter og sekunder
# Først må vi finne ut hvor mange minutter eksperimentet varte totalt
timer = 16
minutter = 16 * 60

# Regn ut hvor mange minutter det var mellom hvert tidssteg, 
# dette tilsvarer den totale tiden i minutter delt på antallet steg. 
steg = minutter/len(temperatur_data)
print("Temperaturmålingene ble gjort hvert " + str(steg) + " minutt")

# Regn ut kun antall minutter - hva tror du int() gjør?
steg_minutter = int(steg)

# Finn residuell fraksjon av minutter, 
# dette er differansen mellom steg og steg_minutter.
steg_differanse = steg - steg_minutter

# Konverter differansen tilbake til minutter:
steg_sekunder = int(steg_differanse * 60)

# Ferdig!
print("Temperaturmålingene ble gjort hvert " + str(steg_minutter) + " minutt og " + str(steg_sekunder) + " sekunder.")
```

Finn en lineær modell som kan passe til dataene du har plottet. 
Bruk den samme metoden som blir brukt for insulinsekresjons-modelleringen i kursboken:

* Bruk det første og siste data-punktet i ditt nye, utvalgte datasett og regn ut stigningstall og skjæringspunkt. Skriv ned funksjonen til modellen.
* Regn ut (simuler) data som beskriver det lineære forholdet mellom stigning og artsrikdom basert på denne modellen.
* Plott de simulerte (beregnede) dataene sammen med alle observerte data, ikke bare dataen du valgte å presentere fra deloppgave b). Husk at en rett linje kan plottes ved å bruke kun to datapunkter.

<details style="background-color: #f9f9f9; padding: 10px; border-radius: 5px; border-left: 4px solid #4CAF50;">

**Stigningstall (a):**

$$
a = \frac{Y_{\text{slutt}} - Y_{\text{start}}}{X_{\text{slutt}} - X_{\text{start}}}
$$

**Konstantledd (b):**

$$
b = Y_{\text{start}} - a \cdot X_{\text{start}}
$$

Får du ikke til å lage den tilpassede lineære modellen, kan du bruke denne modellen i stedet:

$$
y = 0.015x - 7.5
$$
</details>

**Fasit.**
Vi regner ut stigningstallet ved:

$$
a = \frac{y_2 - y_1}{x_2 - x_1}
$$

så hvis vi eksempelvis velger punktene (1300, 11) og (2000, 24) får vi funksjonen:

$a = \frac{24 - 11}{2000 - 1300} = \frac{13}{700} = 0.02$

Vi regner ut skjæringspunktet ved å bruke likningen for en lineær modell, $y = ax + b$, 
velg en verdi for $y$ og løs for $b$. I vårt tilfelle gir det:

$24 = 0.02\cdot2000 + b$

$24 = 40 + b$

$b = 24 - 40 = -16$

Nå har vi stigningstallet $a = 0.02$ og skjæringspunkt $b = -16$, 
dette gir oss den lineære funksjonen $y =  0.02x - 16$

Merk at om du valgte punktene (1300, 11), vil din $b$ bli -15. 
Merk også at alle følgende resultater kan bli litt annerledes hvis du har valgt to ulike datapunkter 
for å regne ut stigningstall og skjæringspunkt.
