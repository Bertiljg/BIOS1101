```python
import pandas as pd
data=pd.read_csv("average-height-by-year-of-birth.csv")
```

```python
print(data)
```

```python
data
```

```python
år=list(data["Year"])
mann_h=list(data["Mean male height (cm)"])
print(mann_h)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(år,mann_h)
plt.xlabel("År")
plt.ylabel("Gjennomsnitshøyde (cm)")
plt.title("Gjennomsnittshøyde for menn med alder 18 år")
plt.show()
```

```python
år_lin = år[0:60]
mann_lin = mann_h[0:60]
```

```python
plt.plot(år_lin, mann_lin)
plt.xlabel("År")
plt.ylabel("Gjennomsnitshøyde (cm)")
plt.title("Gjennomsnittshøyde for menn med alder 18 år mellom 1896-1960")
plt.show()
```

```python
stigning = ((mann_h[60]-mann_h[0])/(år[60]-år[0]))
print(stigning)
```

```python
skjæringspunkt = (mann_h[0] - (stigning*år[0]))
print(skjæringspunkt)
```

```python
print("modellen for den linæere vekstfasen er H =",round(stigning,2), "* t +",round(skjæringspunkt,2))
```

```python
H = 2026 * stigning + skjæringspunkt
print(round(H,2))
```

```python
år_sim = [1896, 2026]
mann_sim = []
```

```python
mann_sim.append(stigning*år_sim[0] + skjæringspunkt)
mann_sim.append(stigning*år_sim[1] + skjæringspunkt)
```

```python
plt.plot(år,mann_h)
plt.plot(år_sim, mann_sim, ":")
plt.title("gjennomsnittshøyde i alder 18 år over tid")
plt.xlabel("År")
plt.ylabel("gjennomsnittshøyde (cm)")
plt.show
```

```python
år_sim = [1850, 2026]
mann_sim = []
mann_sim.append(stigning*år_sim[0]+skjæringspunkt)
mann_sim.append(stigning*år_sim[1]+skjæringspunkt)

plt.plot(år, mann_h, label="Data")
plt.plot(år_sim, mann_sim, ":",label="Linær modell")
plt.xlabel("År")
plt.ylabel("Gjennomsnittshøyde (cm)")
plt.title("gjennomsnittshøyde menn 18 år")
plt.legend()
plt.show
```
