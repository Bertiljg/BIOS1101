## Lineær vekst

Vi bruker

$
H = at + b
$

Der

* $H$ er høyde (i cm)
* $t$ er tid (i år)
* $a$ er stigningstallet, endring i høyde år for år
* $b$ er skjæringspunktet (eller konstantledd), punktet hvor grafen/linjen krysser y-aksen

## Lineær vekst

Vi bestemmer $a$ med

$
a = \frac{y_2-y_1}{t_2-t_1}.
$

Eller i vårt tilfelle

$
a = \frac{H_2-H_1}{år_2-år_1}.
$

<!-- dom:FIGURE: [figures/linear_func.png, width=400] -->
<!-- begin figure -->

<img src="figures/linear_func.png" width="400"><p style="font-size: 0.9em"><i>Figur 1</i></p>
<!-- end figure -->

med hjelp av Python kode:

```python
stigning = (mann[60] - mann[0]) / (år[60] - år[0])
print(stigning)
```

Vi bør skrive dette ut med en forklarende setning:

```python
print("I perioden",år[0], "til", år[60], "ble menn i gjennomsnitt", round(stigning, 2), "cm større per år.")
```

Så beregner vi skjæringspunktet $b$:

For å beregne $b$ bruker vi

$
H = at + b
$

Fra det avleder vi at (sjekk dette!)

$
b = H - at
$

Vi kan velge et hvilken som helst tidspunkt for beregningen.
Her velger vi det første året i datasettet, `år[0]` og `mann[0]`.

```python
skjæringspunkt = mann[0] - stigning * år[0]
print(skjæringspunkt)
```

Vi bør skrive dette ut med en forklarende setning:

```python
print("Modellen for den lineære vekstfasen er H =", round(stigning, 2), "* t +", round(skjæringspunkt,2))
```

Vi trenger en liten korreksjon 😉

```python
print("Modellen for den lineære vekstfasen er H =", round(stigning, 2), "* t", round(skjæringspunkt,2))
```

Hvordan kan vi nå svare på vårt spørsmål "Hvor høyt ville vi vært hadde veksten ikke stoppet opp?"

## Utvide til i år med hjelp av modellen

Høyde i år 2025 i følge modellen:

$
H = at+ b
$

Der

* $t$ = `år[-1]` = $2025$
* $a$ = stigningstall
* $b$ = skjæringspunkt

```python
H = 2025 * stigning + skjæringspunkt
print(round(H, 2))
```

```python
print("Hadde veksten ikke bremset opp ville en gjennomsnittlig mann i 2025 vært", round(H, 2), "cm.")
```

Svar på ett av disse spørsmålene:

Vi overlater svaret til oppgaven a) til dere.

### **b)** Hvor høy var en gjennomsnittlig mann eller kvinne i år 1000?

```python
H = 1000 * stigning + skjæringspunkt
print(round(H, 2))
print("En gjennomsnittlig mann i år 1000 var", round(H, 2), "cm høy.")
```

### **b)** Hvor høy var en gjennomsnittlig mann eller kvinne i år 100?

```python
H = 100 * stigning + skjæringspunkt
print(round(H, 2))
print("En gjennomsnittlig mann i år 100 var", round(H, 2), "cm høy.")
```

Det er god praksis å plotte tallene som modeller gir sammen med datene som 
modellen er bygd på.
Her lager vi nye lister for simulerte tall.

Siden vi for å lage en rett linje i et plott
bare trenger to punkter (startpunktet og sluttpunktet)
bruker vi tidspunkt 1896, og i år.

Vi setter opp to lister, og fyller den for årstall med en gang.

```python
år_sim = [1896,2025]
mann_sim = []
```

Så beregner vi mannlig høyde basert på modellen,
og legger det til listen `mann_sim`

```python
mann_sim.append(stigning * år_sim[0] + skjæringspunkt)
mann_sim.append(stigning * år_sim[1] + skjæringspunkt)
```

Nå kan vi plotte alt sammen:

```python
plt.plot(år, mann)
plt.plot(år_sim, mann_sim, ":")
plt.title("Gjennomsnitshøyde menn i alder 18 år over tid")
plt.xlabel("År")
plt.ylabel("Gjennomsnitshøyde (cm)")
plt.show()
```

Vis er at modellen følger de observte dataen godt,
helt til veksten slutter å være lineært.

Siden vi har flere linjer i plottet,
må vi legge til en tegnforklaring (legend):

```python
plt.plot(år, mann, label="Data")
plt.plot(år_sim, mann_sim, ":", label="Lineær modell")
plt.title("Gjennomsnitshøyde menn i alder 18 år over tid")
plt.xlabel("År")
plt.ylabel("Gjennomsnitshøyde (cm)")
plt.legend()
plt.show()
```

Til slutt kan vi utvide modellen tilbake i tid,
helt til år 0.

Vi legger til blå lije for å vise x aksen.
Til det bruker vi fuksjonen `plt.axhline()`.

```python
år_sim = [0,2025]
mann_sim = []
mann_sim.append(stigning * år_sim[0] + skjæringspunkt)
mann_sim.append(stigning * år_sim[1] + skjæringspunkt)

plt.plot(år, mann, label="Data")
plt.plot(år_sim, mann_sim, ":", label="Lineær modell")
plt.title("Gjennomsnitshøyde menn i alder 18 år over tid")
plt.xlabel("År")
plt.ylabel("Gjennomsnitshøyde (cm)")
plt.legend()
plt.axhline(0)
plt.show()
```
