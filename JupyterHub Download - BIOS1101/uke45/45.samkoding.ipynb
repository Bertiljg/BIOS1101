Den kan vi bruke slik:

```
class DNA:
    def __init__(self, sekvens):
        self.sekvens = sekvens
        self.len = len(sekvens)
        self.bp_vekt = 660 # g/mol

    def hent_vekt(self):
        return self.bp_vekt * self.len

ny_dna = DNA("ACGT")
print(f"Sekvensen veier {ny_dna.hent_vekt()} g/mol.")
```

Den aksepterte måten å gjøre dette på er å lage en metode for det.

```
class DNA:
    def __init__(self, sekvens):
        self.sekvens = sekvens
        self.len = len(sekvens)
        self.bp_vekt = 660 # g/mol

    def hent_len(self):
        return self.len

    def hent_vekt(self):
        return self.bp_vekt * self.len

ny_dna = DNA("ACGT")
print(f"Sekvensen veier {ny_dna.hent_vekt()} g/mol.")
print(f"lengden er {ny_dna.hent_len()}.")
```

## Mer om konstruktøren og metoder

```
class Student:
    def __init__(self, navn, alder, semester):
        self.navn = navn
        self.alder = alder
        self.semester = semester
        
    def hent_navn(self):
        return self.navn
    
    def hent_alder(self):
        return self.alder
    
    def hent_semester(self):
        return self.semester

student1 = Student("Kristine Bonnevie", 20, 1)
print(f"Student {student1.hent_navn()} er {student1.hent_alder()} år.")
student2 = Student("Charles Darwin", 22, 4)
print(f"Min nabo heter {student2.hent_navn()} og er {student2.hent_alder()} år.")

print(f"Min nabo er på {student2.hent_semester()}. semester, mens jeg er på {student1.hent_semester()}. semester.")
```
