## Oppgave 5: Gjeddevekst med objekter

I uke 37 jobbet vi med en oppgave om vekst av en gjedde:

I en innsjø slipper noen ut en gjedde som veier 0.1 kg.
Gjedden forsyner seg grådig av abbor frem til den når 20 kg i vekt.

I oppgaven modellerte vi gjeddens vekst over 20 år,
med utgangspunkt i at den dobler seg i størrelse årlig.
Det tilsvarer en vekstfaktor på 2, eller en vekstrate $r = 1$ og vi brukte en logistisk vekstmodell.

I denne oppgaven skal vi lage et gjedde-objekter som lar oss undersøke gjeddens vekst.

Først må vi gjør om noe av koden vi laget til en funksjon, 
for at den kan bli en metode i klassen vår.

Koden vi brukte for å modellere gjedden vises under:

```
r = 1
K = 20 # K

gjedde_vekter = [0.1]
aar = [0]

teller = 1

while teller <= 20:
    ny_vekt_gjedde = (1 + r * (1 - gjedde_vekter[-1] / K)) * gjedde_vekter[-1]
    gjedde_vekter.append(ny_vekt_gjedde)
    aar.append(teller)
    teller = teller + 1
```
