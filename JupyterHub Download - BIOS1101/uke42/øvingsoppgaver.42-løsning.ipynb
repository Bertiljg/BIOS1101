## Oppgave 2: Writing your own modulo function

> the goal of this exercise is to become familiar with the `modulo` operator as well as practicing building and using functions that perform scientific calculations with `ints` and `floats`.

First, some formal definitions (source: Wikipedia)

> In arithmetic, Euclidean division is the process of division of two integers, which produces a quotient and a remainder.

Example: $11 \div 4 = (2 \times 4) + 3$

Here, `2` is the quotient and `3` the remainder.

> In computing, the *modulo* operation finds the remainder after division of one number by another. Given two positive numbers, $a$ (the dividend) and $n$ (the divisor), $a$ modulo $n$ is the remainder of the Euclidean division of $a$ by $n$.

In the example above, $a = 11$  and $n = 4$ and 11 modulo 4 is 3.

In python we use the *modulo operator* `%`:

```
print(11 % 4)
```

**b)**
Lag en funksjon som teller antall linjer for en hvilken som helst fil.
Input til funksjonen er filnavnet, og funksjonen skal returnere antall linjer i filen.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```
def tell_antall_linjer(filnavn):
    """
    Funksjon som teller antall linjer i en fil
    input: filnavn
    output: antall linjer i filen
    """
    linjeteller = 0
    with open(filnavn, 'r') as fil:
        for linje in fil:
            linjeteller += 1
    return linjeteller


# Kjør funksjonen for å teste
antall_linjer = tell_antall_linjer("hemoglobin_normal.fasta")
print(f"Antall linjer: {antall_linjer}")
```

## Oppgave 6: Sum og gjennomsnitt

Lag en funksjon som tar en liste med tall som input,
og som beregner summen og gjennomsnittet av tallene.
Funksjonen returner **både** summen og gjennomsnittet.
Bruk funksjonen til å skrive en setning om summen og gjennomsnittet av 
en liste med tall.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```
def beregn_sum_og_gjennomsnitt(tall_liste):
    summen = sum(tall_liste)
    gjennomsnitt = summen / len(tall_liste) if tall_liste else 0
    return summen, gjennomsnitt

# Kjør funksjonen for å teste
sum_av_tallene, gjennomsnitt_av_tallene = beregn_sum_og_gjennomsnitt([1, 2, 3, 4, 5])
print(f"Sum: {sum_av_tallene}")
print(f"Gjennomsnitt: {gjennomsnitt_av_tallene}")

# Dette fungerer også:
resultat = beregn_sum_og_gjennomsnitt([1, 2, 3, 4, 5])
print(f"Sum: {resultat[0]}")
print(f"Gjennomsnitt: {resultat[1]}")
```

## Oppgave 7: Fornavn og etternavn

Skriv en funksjon som tar et fullt navn som input (f.eks. "Ola Nordmann")
og returnerer fornavn og etternavn separat.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```
def split_navn(fullt_navn):
    """
    Funksjon som splitter et fullt navn i fornavn og etternavn
    input: fullt navn
    output: fornavn, etternavn
    """
    fornavn, etternavn = fullt_navn.split()
    return fornavn, etternavn

# Kjør funksjonen for å teste
fornavnet, etternavnet = split_navn("Ola Nordmann")
print(f"Fornavn: {fornavnet}")
print(f"Etternavn: {etternavnet}")
```

## Oppgave 8: Partall

Lag en funksjon som et tall som input,
og som returner `True` hvis tallet er et partall, og `False` hvis ikke.
Bruk funksjonen til å skrive ut om tallene 1, 2, 3, 4 og 5 er partall eller ikke.

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```
def er_partall(et_tall):
    """
    Funksjon som sjekker om et tall er et partall
    input: tall
    output: True hvis tallet er et partall, False ellers
    """
    # bruk modulo-operatoren for å sjekke om tallet er partall
    resultat = et_tall % 2 == 0
    return resultat

# Kjør funksjonen for å teste
for tall in [1, 2, 3, 4, 5]:
    if er_partall(tall):
        print(f"{tall} er et partall.")
    else:
        print(f"{tall} er et ikke partall.")
```
