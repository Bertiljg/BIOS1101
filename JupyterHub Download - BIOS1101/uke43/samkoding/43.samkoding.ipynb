## Variabler vs. Lister
Globale variabler: Når du har en global variabel og prøver å oppdatere den inne i en funksjon uten å bruke global-nøkkelordet, oppretter Python en ny lokal variabel med samme navn i funksjonen. Endringene påvirker dermed den lokale versjonen, ikke den globale.

Globale lister (eller muterbare objekter): Lister i Python er muterbare objekter, noe som betyr at innholdet i listen kan endres selv når listen er definert globalt. Når du oppdaterer en liste ved å legge til eller fjerne elementer, modifiserer du objektet i minnet og påvirker dermed den globale listen, selv om du gjør det inne i en funksjon.

## numpy funksjoner for å gi tilfeldige tall:
- random.choice()
- random.random()  # Gir tall mellom 0 og 1!
- random.normal()  # Gir normal distribuerte tall, dvs fra -3 til ca 3

`np.random.random` og `np.random.normal` er begge funksjoner fra NumPy-biblioteket i Python som brukes til å generere tilfeldige tall, men de har forskjellige formål og distribusjoner.

### np.random.random
- **Beskrivelse**: `np.random.random(d)` genererer et array med dimensjoner `d` fylt med tilfeldige flyttall (desimaltall) som er uniformt fordelt mellom 0 og 1. Det betyr at hver verdi har lik sjanse for å bli generert i det intervallet.
- **Eksempel**: Når du bruker `np.random.random(5)`, får du fem tilfeldige tall som kan være alt fra 0.0 til 1
de to funksjonene!

### np.random.normal
- **Beskrivelse**: `np.random.normal(loc, scale, size)` genererer tall som følger en normalfordeling (Gaussisk fordeling). Her er `loc` forventningsverdien (mean) og `scale` standardavviket.
- **Eksempel**: Når du bruker `np.random.normal(5, 1, 1000)`, får du 1000 tilfeldige tall som er fordelt rundt 5 (med et standardavvik på 1). Dette betyr at de fleste tallene vil være nærmere 5, men noen vil være lavere og noen høyere, med sjeldnere ekstremverdier.
