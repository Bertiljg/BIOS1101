````
import matplotlib.pyplot as plt
N = 180         # number of time steps
r = 0.035       # growth rate

E = []
t = []

# Initial conditions
E.append(10042)   # initial number of bacteria
t.append(60)      # initial measurement time (in minutes)

# Calculating number of bacteria and time for each generation
n = 1
while n < N:
    E_n = (1 + r) * E[n-1]
    E.append(E_n)

    t_n = t[n-1] + 1
    t.append(t_n)

    n = n + 1

# Plotting the modeled data
plt.plot(t, E, label="Modeled")
plt.xlabel("Time, t (minutes)")
plt.ylabel("Population size, E")
plt.title("Modeled bacterial population growth")
plt.legend()
plt.show()
````

## Oppgave 9: A simple way to add vaccination to the SIR model

In the course book, a so-called SIR model for epidemics is introduced in chapter 9.
Below is the full program from the chapter (section 9.3)
that was used to generate figure 9.6 in the book.
However, two things have changed to make it a little bit easier
to answer the questions in this exercise:

* the number of people in the village has been set to 501
* only one of them is infected at the start of the model

An important aspect of the SIR model is that it divides the population
into discrete compartments (or categories):

* Susceptible: healthy people that can become infected
* Infected: these can also infect others
* Recovered: these can not be infected, nor infect others

````
import matplotlib.pyplot as plt
import numpy as np
# Set parameters
N = 24        # number of weeks
a = 0.0027    # infection rate
b = 0.25      # recovery rate

# Create the arrays to store the number of individuals in each group
t = np.arange(N)
S = np.zeros(N)
I = np.zeros(N)
R = np.zeros(N)

S[0] = 500 # Susceptible
I[0] = 1   # Infected
R[0] = 0   # Recovered

for n in range (1, N):
    S[n] = S[n-1] - a*S[n-1]*I[n-1]
    I[n] = I[n-1] + a*S[n-1]*I[n-1] - b*I[n-1]
    R[n] = R[n-1] + b*I[n-1]

plt.plot(t, S, "-*", label = "Susceptible")
plt.plot(t, I,"-*", label = "Infected")
plt.plot(t, R,"-*", label = "Recovered")
plt.xlabel("Weeks")
plt.ylabel("Number of people")
plt.legend()
plt.show()
````

## Den reverse komplementære DNA-strengen

Den doble heliksen av DNA fungerer slik at nukleotidene i hver DNA-streng ikke er helt like, men er i stedet "komplementære" tråder med komplementære nukleotidbaser. Det vil si at hvert nukleotid vil alltid matche med en annen spesifik type nukleotid på den motsatte strengen. Adenin (A) vil alltid bli komplementert med tymin (T), og cytosin (C) alltid med guanin (G).

Når vi jobber med en DNA sekvens, så ser vi ofte på kun en av strengene. "Revers komplementstrengen" er sekvensen til den andre strengen og skiller seg fra partneren sin ved at:

* A erstattes med T, T med A, G med C, og C med G
* rekkefølgen av nukleotidene går baklengs

Således er den reverse komplementære strengen til `ACCTGAG` = `CTCAGGT`
