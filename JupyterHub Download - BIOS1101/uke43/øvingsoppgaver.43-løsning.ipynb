## Oppgave 1: Differensligning (3)

Du er gitt denne differensligningen:

$x_n = 1.5 \times x_{n-1} - 200$

Med begynnverdien

* $x_0 = 2000$

Du har muligens tidligere løst denne oppgaven med hjelp av en while og/eller en for løkke.

Skriv et python program som
* beregner $x_n$ for $n$ fra og med 0 til og med 20 med hjelp av en løkke *og Numpy arrays*
* lager en graf med $x$ (på y-axis) versus $n$ (på x-axis), inkludert navn på akser ('axis label') og tittel på plottet

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**
Generell:
* I løsningsforslagene antas at $x$ verdier lagres i en liste med navn `x` og $n$ i en liste `n`, men andre navn er OK også
* I løsningsforslagene under bygges listen `n` opp i løkken, men det er også OK med `n = range(21)` eller `n = arange(21)`

Løsning med Numpy Arrays

````
import matplotlib.pyplot as plt
import numpy as np
x = np.zeros(21)
x[0] = 2000
n = np.zeros(21)

for i in range(1,21):
    x[i] = x[i-1] * 1.5 - 200
    n[i] = i

plt.plot(n, x)
plt.title("min differensligning")
plt.xlabel("n")
plt.ylabel("x")
plt.show()
````

Løsning med vanlige lister, while-løkke, tas med her også:

````
import matplotlib.pyplot as plt
x = []
x.append(2000)
n = []
n.append(0)

i = 1

while i < 21:
    x.append(x[i-1] * 1.5 - 200)
    n.append(i)
    i = i + 1

plt.plot(n, x)
plt.title("min differensligning")
plt.xlabel("n")
plt.ylabel("x")
plt.show()
````
