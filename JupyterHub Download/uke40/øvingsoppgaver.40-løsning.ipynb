## Oppgave 7: PCR

Polymerase Chain Reaction (PCR) is a technique used in molecular biology to amplify a fragment of DNA. PCR can be used to amplify small amounts of DNA to much larger amounts, sufficient to do further molecular biological experiments. PCR is usually performed for a number of cycles. In each cycle, a new copy of each target DNA is made. If we assume a PCR is 100% efficient it *doubles* the number of target molecules every cycle. However, this is normally not the case. There are many factors, such as the total amount of reagents, that limit the efficiency of the reaction during different steps. 
A more realistic efficiency for a PCR reaction is 60%, meaning that, for each cycle, *60% of the original DNA gets copied and 20% does not*

For this exercise, you'll need these formulas:

* The formula for calculating the number of DNA molecules given a certain amount of DNA is

> Number of DNA molecules = (weight of DNA in gram \* Avogadro's number) / (length of DNA in bases \* 650 gram/mole)

* The formula for converting the number of DNA molecules of DNA into their weight is

> weight of DNA in gram = (number of DNA molecules \* length of DNA in bases * 650 gram/mole) / Avogadro's number

with Avogadro's number = $6.022\times10^{23}$

For the PCR reaction:
* 1 picogram, or $1\times10^{-12}$ gram, of the DNA to be amplified will be present at the beginning of the reaction
* the length of DNA to be amplified, in bases, is 1666
* the PCR will be performed using 30 cycles

Write a Python program that:
* calculates the number of copies of the target after the 30 cycles have ended, assuming 80% PCR efficiency, using a for loop
* calculates the final weight of amplified target DNA in nanograms, $1\times10^{-9}$ gram

<!-- --- begin answer of exercise --- -->
**Fasit.**

````
weight = 1e-12 # 1 picogram
avogadro = 6.022e23
genome = 1666
````

Calculate number of starting molecules

````
molecules = (weight * avogadro) / (genome * 650)
print(molecules)
````

Or, shorter but less readable:

````
molecules = (1e-12 * 6.022e23) / (1666 * 650)
print(molecules)
````

Calculate number of molecules after PCR

Using a loop:

````
import numpy as np
copies = np.zeros(31)
copies[0] = molecules

for cycle in range(1, 31):
  copies[cycle] = copies[cycle - 1] * 1.6
final_copies = copies[-1]
print(final_copies)
````

Using direct calculations: final number of copies is $molecules\times1.6^{30}$

````
final_copies = molecules*1.6**30
print(final_copies)
````

Calculate final weight:

````
final_weight = ((final_copies * 1666 * 650) /  6.022e23) * 1e9
print(final_weight)
````
