```python
# %load code/logist.py
import matplotlib.pyplot as plt

xs = []
ns = range(0, 26)

xs.append(0.1)

r = 0.3
K = 1

n = 1
while n < 26:
    x_n = (1 + r * (1 - xs[n-1]/K))*xs[n-1]
    xs.append(x_n)
    n = n + 1

plt.plot(ns, xs)
plt.title("Logistisk vekst")
plt.xlabel("xs")
plt.ylabel("ns")
plt.axhline(K, ls = ":")
plt.show()
```

```python
rs = [0.1, 0.3, 1, 1.5, 2]
K = 1

i = 0
while i < len(rs):

    xs = []
    ns = []

    xs.append(0.1)
    ns.append(0)

    r = rs[i]

    n = 1
    while n < 26:
        x_n = (1 + r * (1 - xs[n-1]/K))*xs[n-1]
        xs.append(x_n)
        ns.append(n)

        n += 1
    plt.plot(ns, xs, label = f"r = {r}")

    i += 1


plt.axhline(K, ls = ":")
plt.title("Logistisk vekst med ulike r-verdier")
plt.xlabel("Tidssteg")
plt.ylabel("Populasjonsstørrelse")
plt.legend()
plt.show()
```

```python
Ks = [1, 5, 10, 50, 100]
r = 0.3

i = 0
while i < len(Ks):
    xs = []
    ns = []

    xs.append(0.1)
    ns.append(0)

    K = Ks[i]

    n = 1
    while n < 40:
        x_n = (1 + r * (1 - xs[n-1]/K))*xs[n-1]
        xs.append(x_n)
        ns.append(n)
        n = n + 1
    plt.plot(ns, xs, label = f"K = {K}")
    plt.axhline(K, ls=":")

    i = i + 1
plt.title("Logistisk vekst med ulike K")
plt.yscale("log")
plt.xlabel("Tidssteg")
plt.ylabel("populasjonsstørrelse")
plt.legend()
plt.show()
```
