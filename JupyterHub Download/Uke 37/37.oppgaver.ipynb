## Oppgave 1: Logistisk vekst av havørn

Logistisk vekst av en havørn populasjon kan modelleres med denne
logistiske vekstmodellen:

$$
P_t = P_{t-1} + r \cdot P_{t-1} \cdot \left(1 - \frac{P_{t-1}}{K}\right)
$$

hvor:
* $P_t$ er bestanden ved tid $t$
* $r$ er vekstrate
* $K$ er bæreevnen (det maksimale antallet individer miljøet kan støtte)

Her er kodelinjer som, satt sammen i riktig rekkefølge, utgjør et program
som simulerer og plotter veksten av havørnbestanden over tid.
Merk at noen av linjene ikke skal være en del av programmet.

```python
import matplotlib.pyplot as plt

P0 = 10  # Startbestand
K = 500  # Bæreevne
r = 0.1  # vekstrate
år = 1
t = [0]
P = [P0]

while år <= 100:
    P.append(P[år-1] + r * P[år-1] * (1 - P[år-1] / K))
    t.append(år)
    år = år + 1

#t = t + 1
#P.append(P[år] + r * P[år] * (1 - P[år] / K))
#t = 0

plt.plot(t, P)
plt.xlabel('Tid (år)')
plt.title('Logistisk vekstmodell av havørn')
plt.ylabel('Havørnbestand')
plt.show()
```

Det er ingen andre predator fisk i vannet før gjedda sleppes ut.
Det antas at gjedda kun spiser av abboren og at det derfor kun kan bli 20 kg gjedde i vannet.
Videre antar vi at gjedda dobbler seg i vekt for hvert år som går.

**a)**
Vi skal nå beregne veksten for gjedda i vannet ved bruk av en logistisk vekst modell.

$$
G_n  = \left(1 + r \left( 1- \frac{G_{n-1}}{K} \right) \right)G_{n-1}.
$$

Der
* $G$ er gjeddes vekt
* $n$ er årene
* $r$ er vekstraten
* $K$ er maksimal gjeddevekt i kilogram

Lag et Python program som beregner hvordan vekten av gjedden utvikler seg i løpet av 20 år.


```python
import matplotlib.pyplot as plt

P0 = 0.1  # Startvekt
K = 20  # Bæreevne
r = 1  # vekstrate
år = 1
t = [0]
P = [P0]

while år <= 20:
    P.append((1+ r * (1 - P[år-1] / K))* P[år-1])
    t.append(år)
    år = år + 1
print(P)
```

```python
plt.plot(t, P)
plt.xlabel('Tid (år)')
plt.ylabel('Vekten til en ensom gjedde (Frank)')
plt.axhline(K, ls=":", color="red")
plt.title('Logistisk vekstmodell av en ensom gjedde (Frank)')
plt.show()
```
