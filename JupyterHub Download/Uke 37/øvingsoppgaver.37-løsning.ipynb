## Oppgave 4: f-strenger

Kopier programmet under og endre det slik at print kommandoer bruker f-strenger:

```
temp_i_gaar = 19
print("Temperaturen i går var", temp_i_gaar, "grader.")

navn = "Puss"
alder = 4
print("Katten heter "+ navn + " og er " + str(alder) + " år gammel.")

fag = ["BIOS1101", "BIOS1110", "BIOS1150"]
print("Du har", len(fag), "i studieprogrammet ditt.")

a = 1.25
b = -1
print("Beregningen gir", round(a**3 + b, 2), "som det endelige svaret.")
```

<!-- --- begin solution of exercise --- -->
**Løsningsforslag.**

```
temp_i_gaar = 19
print(f"Temperaturen i går var {temp_i_gaar} grader.")

navn = "Puss"
alder = 4
print(f"Katten heter {navn} og er {alder} år gammel.")

fag = ["BIOS1101", "BIOS1110", "BIOS1150"]
print(f"Du har {len(fag)} i studieprogrammet ditt.")

a = 1.25
b = -1
print(f"Beregningen gir {a**3 + b:.2f} som det endelige svaret.")
```
