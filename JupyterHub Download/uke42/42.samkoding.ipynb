## Lese FASTA filer med open funksjonen

Nå skal vi lære en annen måte å åpne filer og hente inn data fra dem på.

Python har en funksjon `open()` for å åpne en hvilken som helst

````
fh = open("hemoglobin_normal.fasta", "r")
````

Verdt å legge merke til:
* det første argumentet for `open()` er navnet på filen vi ønsker å åpne gitt som en streng
* det andre argumentet, strengen `"r"`, (for 'read', å lese), forteller at vi ønsker å åpne filen for lesing
* ofte brukes `fh` som variabelnavn sammen med `open()`;  dette står for 'filehandle'

Innholdet i filen kan leses på forskjellige måter,
men en vanlig metode er å bruke funksjonen `readlines()`.
Denne funksjonen returnerer en liste
der hver linje i filen vår er et element i listen.

````
linjer = fh.readlines()
print(linjer)
````

Disse dataene lagres i et såkalt 'FASTA'-format.
FASTA er et populært filformat som brukes til lagring av DNA-sekvenser.
Den første linjen starter med et `>` (større enn symbol) og inneholder
navnet og noen ganger forskjellige andre ting om den spesifikke sekvensen.
Denne linjen kalles en 'header', som kan oversettes med 'overskrift'.
Våre mRNA-hemoglobinsekvensfiler har følgende overskrift:

```
        >NM_000518.4 Homo sapiens hemoglobin subunit beta (HBB), mRNA
```

Overskriften forteller oss at vi har mRNA for en spesifikk
subunit av det menneskelige hemoglobin genet.

Hver linje i vi får med `print(linjer)` over slutter med tegnene `\n`.
Dette er *newline* -symbolet og betyr en linjeskift:
neste tekst kommer på neste linje.
Når vi skriver ut en streng, blir symbolet `\n` gjort om til en linjeskift
og teksten fortsetter på neste linje:

````
print("Hei, verden!")
print("Hei\nverden!")
````

Vi bør alltid lukke en fil når vi er ferdig med å jobbe med den.
Dette frigjør ressursene datamaskinen brukte for å åpne den,
og bidrar til å forhindre tap av data.
Dette gjør vi ved å bruke `close()`:

````
fh.close()
````

Per nå er DNA strenger splittet over flere strenger som danner elementene i
listen vår.
Vi ønsker å jobbe med en sammensatt streng som inneholder hele mRNA-sekvensen.
Vi gjør `linjer` til en streng ved å starte med en tom streng som heter `mRNA`.
Så lager vi en løkke som går over hvert element av `linjer`
og legger dem til` mRNA` strengen.
Hvert element i `linjer` er en streng og å legge sammen strenger gjøres med `+`.
Etter løkken skriver vi ut resultatet:

````
mRNA = ""
for linje in linjer:
    mRNA += linje

print(mRNA)
````

Selv om selve DNA strengen nå er i en variabel,
så ser vi at når vi skriver den ut at den går over flere linjer.

Her bruker vi `.strip()` for å fjerne linjeskift-symbolene fra hver linje i `linjer`.

### Tilbake fil FastQ filen

````
for line in lines:
    print(line.strip())
````

Vi ønsker å hente ut sekvensene, altså informasjonen som er på hver 2. linje.

Siden mønsteret gjentar seg for hver fjerde linje,
kan vi bruke resten ved divisjon på `4`
til å se hva slags linje vi står på.

````
for i in range(len(lines)):
    print(i, i % 4, lines[i].strip())
````

Legg merke til:

* `i % 4 == 0` betyr header-linje
* `i % 4 == 1` betyr sekvenslinje
* `i % 4 == 2` betyr `+`-linje
* `i % 4 == 3` betyr kvalitetslinje

Vi kan derfor lage en funksjon som henter ut alle sekvensene:

````
def load_fastq_sequences(filename):
    """
    Read a FASTQ file and return the sequences as a list.
    """
    fh = open(filename, "r")
    lines = fh.readlines()
    fh.close()

    sequences = []

    for i in range(len(lines)):
        line = lines[i].strip()

        if i % 4 == 1:
            sequences.append(line)

    return sequences
````

Et eksempel:

````
sequences = load_fastq_sequences("example_reads.fastq")
print(sequences)
````

<div class="alert alert-block alert-warning">

<h2>Vis lysbilde(r):</h2>

* Fra FASTQ til FASTA



</div>

Nå vil vi gjøre disse sekvensene om til `FASTA` format.

Da lager vi en ny funksjon som returnerer én streng
med hele `FASTA` teksten:

````
def fastq_to_fasta_text(filename):
    """
    Read a FASTQ file and return the sequences as FASTA text.
    """
    sequences = load_fastq_sequences(filename)

    fasta_text = ""
    read_number = 1

    for sequence in sequences:
        header = f">read_{read_number}\n"
        fasta_text += header
        fasta_text += sequence + "\n"
        read_number += 1

    return fasta_text
````

Et eksempel:

````
fasta_text = fastq_to_fasta_text("example_reads.fastq")
print(fasta_text)
````

### Kan vi bruke open for csv filer?

Vi har brukt `open()` for å lese inn data fra en FASTA fil.
Men før det brukte vi altid `pandas` og `pandas.read_csv` til å lese inn
data fra `csv` filer.

Kunne vi brukt `open()` til å lese csv filer også?

Svaret er ja:

````
fh = open("insulin_prod_non_obese.csv", "r")
linjer = fh.readlines()
print(linjer)
````

Verdt å legge merke til:
* vi får igjen hver linje som et element i en liste
* disse elementene er strenger, med linjeskift symbolet på slutten
* det står en komma mellom dataene som i csv filen, dataene er ikke splittet i kolonner

For å få dataene splittet i kolonner og samle hver kolonne i en liste
som vi får med bruk av `pandas`, må vi gjøre noen flere ting:
* fjerne linjeskift
* splitte på komma
* gjøre om til float (`readlines()` gir strenger)
* legge til lister

Her demonstreres hvordan det kan gjøres:

````
fh = open("insulin_prod_non_obese.csv", "r")
linjer = fh.readlines()

# lag to nye lister
ts = []
Ns = []

# løkker over alle linjer untatt første, som er overskriftene
for linje in linjer[1:]:
    kolonner = linje.strip() # fjern linjeskift
    splittet = kolonner.split(',') # splitt på komma
    ts.append(float(splittet[0])) # første kolonne er ts, gjør om til float og legg til
    Ns.append(float(splittet[1])) # andre kolonne er Ns, gjør om til float og legg til

print(ts)
print(Ns)
````

Dette gir samme resultat, men er mer omfattende enn `pandas` løsningen
som vi dermed foretrekker for data i `csv` format.
