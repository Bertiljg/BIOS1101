## Oppgave 8: Sommer

En tekstfil "sommer.txt" har følgende innhold :

        Month,Rainfall
        Average,19.3
        June,20.0
        July,10.5
        August,27.5


Hva skrives ut når følgende kode kjøres? Du kan anta at koden kjører uten
feilmeldinger.

````
import pandas as pd
data = pd.read_csv('sommer.txt')
month = list(data["Month"])
rainfall = list(data["Rainfall"])

total_rain = 0
i = 0

while i < len(month):
    if month[i] in ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]:
        total_rain += rainfall[i]
    else:
        print(f"Found a non-month in the data: {month[i]} with rainfall {rainfall[i]}")
    i += 1
print(f"Total rain in the summer: {round(total_rain, 1)} mm.")
````

## Oppgave 24: Fibonacci

Hva skrives ut når dette programmet kjøres?

````
import numpy as np

def fibonacci(N):
    x = np.zeros(N+1)
    x[0] = 1
    x[1] = 1
    for n in range(2, N+1):
        x[n] = x[n-1] + x[n-2]
    return x

print(max(fibonacci(5)))
````
