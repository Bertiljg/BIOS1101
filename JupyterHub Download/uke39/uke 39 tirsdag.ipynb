```python
# %load code/mendel.py
import numpy as np, matplotlib.pyplot as plt

plante1_p = ["P", "p"]
plante2_p = ["P", "p"]
hvit = 0
N = 1000

frekvens = []
krysninger = []

n = 1
while n < N+1:
    allel_plante1 = np.random.choice(plante1_p)
    allel_plante2 = np.random.choice(plante2_p)
    genotype = allel_plante1 + allel_plante2
    if genotype == "pp":
        hvit = hvit + 1
    frekvens.append(hvit/n)
    krysninger.append(n)
    n = n + 1


#print(f"frekvensen av hvite blomster er {frekvens} og antall krysninger er {krysninger}")

plt.plot(krysninger, frekvens, "-g", label = "Frekvens av hvite blomster")
plt.xlabel("frekvens")
plt.ylabel("antall krysninger")
plt.title("Frekvens av hvite blomster som funksjon av antall krysninger")
plt.legend()
plt.show()
```
