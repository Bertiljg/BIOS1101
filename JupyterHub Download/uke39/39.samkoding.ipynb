## Beslutninger og if-setninger

Du skal til UiO og *avhengig av* om du har tatt med matpakken må du handle på Bunnpris.
Med andre ord: du må *ta en beslutning* om du skal handle eller ikke.

I pseudokode:

        hvis jeg ikke har med matpakke:
          jeg må handle på Bunnpris


Vi skal bruke akkurat samme type beslutning i snegleprosjektet:

        hvis skallengden er mindre enn 21 mm:
          sneglen er liten


Slike beslutninger gjør man i Python med såkalte *if-setninger*.

```python
terskel_mm = 21

en_lengde = 19

if en_lengde < terskel_mm:
    print(f"Sneglen med lengden {en_lengde} mm klassifiseres som 'liten'.")

print("Ferdig.")
```

Her er `if` en beslutningsregel.

Ting å legge merke til:

* selve `if`-setningen ligner mye på en `while`-setning
* setningen avsluttes med en kolon `:`
* `if`-setningen etterfølges av en kodeblokk som er innrykket
* vi sier at `if`-setningen sjekker om en betingelse er oppfylt, eller er sann
* bare hvis betingelsen er oppfylt kjøres den innrykkede kodeblokken
* hvis betingelsen ikke er oppfylt, hopper Python over den innrykkede kodeblokken

Hva skjer hvis vi har en stor snegle?

```python
terskel_mm = 21

en_lengde = 23

if en_lengde < terskel_mm:
    print(f"Sneglen med lengden {en_lengde} mm klassifiseres som 'liten'.")

print("Ferdig.")
```

kjøres ikke kodelinjen under den.

## En funksjon for sampling

Vi lager også en funksjon som gjør selve sampling-steget:

```python
def tilfeldig_snegl(snegle_lengder):
    sample = np.random.choice(snegle_lengder)
    return sample
```

Test funksjonen et par ganger:

## To funksjoner som samarbeider

Ofte er det lurt at hver funksjon har en liten, tydelig jobb.

* `klassifiser_snegl_lengde()` tolker observasjonen

Nå lager vi et program som bruker begge:

```python
en_lengde = tilfeldig_snegl(lengdene)
lengde_klasse = klassifiser_snegl_lengde(en_lengde)

print(f"Observasjon: en snegle med lengde {en_lengde} mm.")
print(f"Klassifiering: {lengde_klasse}.")
```

## Gjøre terskelverdien til en parameter

Vi har kodet terskelverdien inn i funksjonen med linjen `terskel_mm = 21`.
Om vi ønsker å sammenligne forskjellige terskelverdier
er det smarterer at vi kan ha den med som parameter:

```python
def klassifiser_snegl_lengde(lengde_mm, terskel_mm):
    if lengde_mm < terskel_mm:
        klasse = "liten"
    else:
        klasse = "stor"
    return klasse

terskel = 21

en_lengde = tilfeldig_snegl(lengdene)
lengde_klasse = klassifiser_snegl_lengde(en_lengde, terskel)

print(f"Observasjon: en snegle med lengde {en_lengde} mm.")
print(f"Klassifiering: {lengde_klasse}.")
```

Nå har vi en funksjon med to parametrene.

## Mange observasjoner med while-løkke

Én observasjon er ikke nok.
Nå vil vi sample flere snegler og telle hvor mange som havner i hver kategori.

Vi lager et program som gjør dette med en `while`-løkke:

```python
N = 10 # antall snegl vi trekker fra datasettet
terskel = 21

antall_liten = 0
antall_stor = 0

n = 0
while n < N:
    en_lengde = tilfeldig_snegl(lengdene)
    lengde_klasse = klassifiser_snegl_lengde(en_lengde, terskel)

    if lengde_klasse == "liten":
        antall_liten = antall_liten + 1
    else:
        antall_stor = antall_stor + 1

    n = n + 1

print(f"Vi observerte {antall_liten} små og {antall_stor} store snegler.")
print(f"Andel små snegler: {100*antall_liten/N:.0f}%")
```

## Et stolpediagram av resultatet

Til det bruker vi `pyplot` funksjonen `plt.bar`.

```python
kategorier = ["stor", "liten"]
antall = [antall_liten, antall_stor]
plt.bar(kategorier, antall, color=["green", "blue"])
plt.xlabel("Klasse")
plt.ylabel("Antall snegler")
plt.title(f"Utvalg av {N} snegler")
plt.show()
```

Det er mye fram og tilbake mellom kodeblokkene. 
Vi kunne laget en stor kodeblokk, men velger heller å
gjøre om plottet til en funksjon også:

```python
def plot_size_antall(antall_liten, antall_stor, title):
    kategorier = ["stor", "liten"]
    antall = [antall_liten, antall_stor]
    plt.bar(kategorier, antall, color=["green", "blue"])
    plt.xlabel("Klasse")
    plt.ylabel("Antall snegler")
    plt.title(f"Utvalg av {N} snegler")
    plt.show()
```

Legg merke til at denne funksjonen har tre parametrene, men ingen returverdi!

## Få snegler: første sampling

Vi kjører igjen med et utvalg på 10:

```python
N = 10 # antall snegl vi trekker fra datasettet
terskel = 21

antall_liten = 0
antall_stor = 0

n = 0
while n < N:
    en_lengde = tilfeldig_snegl(lengdene)
    lengde_klasse = klassifiser_snegl_lengde(en_lengde, terskel)

    if lengde_klasse == "liten":
        antall_liten = antall_liten + 1
    else:
        antall_stor = antall_stor + 1

    n = n + 1

plott_tittel = f"Sample of {N} snails"
plot_size_antall(antall_liten, antall_stor, plott_tittel)
```

Dette plottet viser bare ett tilfeldig utvalg på 10 snegler.
Det betyr at plottet kan se ganske annerledes ut neste gang vi kjører samme kode.
