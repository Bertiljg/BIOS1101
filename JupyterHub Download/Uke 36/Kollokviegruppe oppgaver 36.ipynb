her skal vi bruke differenslikningen for å finne punktet hvor arealet er halveis dekt av alger.



økningen i areal er *2 per dag

Starten av algevekst:

 t_0 = 1.juni

Slutt av algevekst:

t_max = 30 juli.

### a: Hvilken dag er halve innsjøen dekket?

30 dager i juni

31 dager i juli

= 61 dager totalt

y = 0 -> 100% dekket

x = tid i dager, gitt ved 1 -> 60



```python
import numpy as np
import matplotlib.pyplot as plt

dag = 1
L_dag = list(range(1,61))
L_areal = []

while dag < 61:
    L_areal.append(100*2**(dag-60))
    dag = dag + 1

print("Innsjøen er 50% dekt på dag:", L_areal.index(50)+1)
plt.plot(L_dag, L_areal)
plt.xlim(53,60)
plt.axhline(50, ls=':')
plt.axvline(L_dag[L_areal.index(50)], ls=':')
plt.xlabel("dager, tid")
plt.ylabel("areal dekt")
plt.show()
```

#### 1.
Dette er en differenslikning som beskriver en størrelse i en tallrekke ved en gitt indeks gitt man kjenner den forrige mengden og endringsraten mellom dem.

#### 2.
Vekstraten mellom de to stegene i en differenslikning.

#### 3.
$r = 1$ siden arealet dekket av alger dobles per dag.

arealet for en gitt tid er lik vekstraten multiplisert med forrige tidssteg. Hvis arealet skal dobles per steg må (1 + r) være lik 2.

```python
import numpy as np
import matplotlib.pyplot as plt

l_areal_c = []
l_dager_c = []
dager_c = 1
areal_c = 1.8*10**-8
#utgangsverdier
#start 1.8*e-8
#slutt 10 000

while areal_c < 10000:
    l_areal_c.append(areal_c)
    l_dager_c.append(dager_c)
    dager_c = dager_c + 1
    areal_c = areal_c*2

l_areal_c.append(areal_c)
l_dager_c.append(dager_c)

print("Innsjøen vil være dekt på dag:", l_dager_c[-1])
plt.plot(l_dager_c, l_areal_c)
plt.yscale("log")
plt.axhline(10000, ls=':', color="red")
plt.xlabel("Dager")
plt.ylabel("Areal i kvadratkilometer")
plt.show()
```
