Programsetningen `indeks = indeks + 1` kan virke merkelig
hvis du ser på det fra et matematisk synspunkt.
Hvordan kan `indeks` være det samme som `indeks + 1`?

Men i Python betyr `=` ikke "er lik" i matematisk forstand.
Det betyr at variabelen til venstre får tildelt verdien som beregnes på høyre side.
Dermed kan en variabel godt stå på begge sider av `=`.

Hvis `indeks` for eksempel er `2`,
så blir `indeks + 1` først regnet ut til `3`,
og deretter får variabelen `indeks` den nye verdien `3`.

## Introduksjon av while-løkker

Nå skal vi automatisere denne prosessen med en `while` løkke.
"While" på engelsk betyr her "så lenge som".
Du kan tenke på en while-løkke slik:
"Så lenge vi har mer å gjøre, gjør det".

<div class="alert alert-block alert-warning">

<h2>Vis lysbilde(r):</h2>

* Struktur til while løkker



</div>

Slik blir vår løkke:

````
lengder = []
indeks = 0

while indeks < 4:
    sekvens = dna_sekvenser[indeks]
    lengde = len(sekvens)
    lengder.append(lengde)
    indeks = indeks + 1

print(lengder)
````

<div class="alert alert-block alert-info">
Forklar løkken steg for steg.
Pek spesielt på:

* initiering: `indeks = 0`
* betingelse: `indeks < 4`
* kodeblokken inni løkken
* oppdatering: `indeks = indeks + 1`

</div>

Legg merke til disse tingene:

* linjen med `while` slutter med kolon `:`
* koden inni løkken er innrykket
* i Jupyter Notebook får du innrykk automatisk når du avslutter en linje med `:`
* det er viktig at alle linjer inni kodeblokken har like mye innrykk

<div class="alert alert-block alert-info">
Mulig forvirring:
kolon `:` på slutten av `while` linjen er ikke det samme som kolon
i slicing, for eksempel `liste[1:3]`.
</div>

Vi sier at en while-løkke består av en *betingelsestest*
og en *kodeblokk*.
Linjen `indeks < 4` er en betingelse.
En slik betingelse evalueres til enten `True` eller `False`.

Vi kan se på dette direkte:

````
indeks = 0
indeks < 4
````

Her får vi `True`,
fordi `0` faktisk er mindre enn `4`.

Hva om `indeks` er `4`?

````
indeks = 4
indeks < 4
````

Da får vi `False`,
fordi listen vår bare har indeksene `0`, `1`, `2` og `3`.

Vi kan også få løkken til å skrive ut litt informasjon underveis:

````
lengder = []
indeks = 0

while indeks < 4:
    print("Når indeks er", indeks, "er betingelsen", indeks < 4)
    sekvens = dna_sekvenser[indeks]
    lengde = len(sekvens)
    lengder.append(lengde)
    indeks = indeks + 1

print("Løkken er ferdig.")
print("Når indeks er", indeks, "er betingelsen", indeks < 4)
print(lengder)
````

Det er nyttig å merke seg at vi her bygger opp en ny liste litt etter litt.
Det er ofte det løkker brukes til:
gjenta en liten oppgave mange ganger,
og samle resultatene underveis.

### Vanlige feil når man lager en while løkke

En av de vanligste feilene er å glemme kolon på slutten av `while` linjen:

````
lengder = []
indeks = 0
while indeks < 4
    sekvens = dna_sekvenser[indeks]
    lengder.append(len(sekvens))
    indeks = indeks + 1
````

Dette fører til en `SyntaxError`.

En annen vanlig feil er å glemme å endre verdien av løkkevariabelen:

```Python
        lengder = []
        indeks = 0
        while indeks < 4:
            sekvens = dna_sekvenser[indeks]
            lengder.append(len(sekvens))
```

Da får vi en *uendelig løkke*,
fordi `indeks` aldri blir større.
Notebooken fortsetter da å jobbe helt til vi avbryter den.

## En mer robust while løkke

Løsningen over virker,
men den har en svakhet:
vi måtte endre selve løkken da listen fikk ett nytt element.

Det er ofte et tegn på at koden kan forbedres.
Vi ønsker helst en løkke som automatisk tilpasser seg
hvor lang listen faktisk er.

Da kan vi bruke `len(dna_sekvenser)` i betingelsen:

````
lengder = []
indeks = 0

while indeks < len(dna_sekvenser):
    sekvens = dna_sekvenser[indeks]
    lengde = len(sekvens)
    lengder.append(lengde)
    indeks = indeks + 1

print(lengder)
````

Her betyr `len(dna_sekvenser)` "lengden til listen `dna_sekvenser`".
Hvis listen har 4 elementer blir betingelsen det samme som `indeks < 4`.
Hvis listen har 5 elementer blir den det samme som `indeks < 5`.

Dette er en mer robust løsning,
fordi den fortsatt virker selv om vi endrer på listen.

## Lese inn de virkelige dataene

Merk: dette er en av oppgavene fra forrige uke.

Dataene ligger i filen `antall-meldte-covid-19-t.csv`.
Vi leser dem inn med `pandas` og gjør kolonnene om til lister:

````
import pandas as pd

tilfeller = pd.read_csv("antall-meldte-covid-19-t.csv")
print(tilfeller)

datoer = list(tilfeller["Dato"])
antall = list(tilfeller["Kumulativt antall"])
nye = list(tilfeller["Nye tilfeller"])
````

Nå skal vi bruke en `while` løkke til å beregne dag for dag
helt til vi når målet.

While løkken kan beskrives som "så lenge vi har flere å smitte, fortsett".

````
antall_smittet = 1
modell = [antall_smittet]
dager_modell = [0]

R0 = 1.5
maal = 5600000
dag = 0

while antall_smittet < maal:
    antall_smittet = antall_smittet * R0
    dag = dag + 1
    modell.append(antall_smittet)
    dager_modell.append(dag)

print("Det går", dag, "dager før hele Norges befolkning er smittet i modellen")
print("Da er", round(antall_smittet), "mennesker smittet")
````

Her bruker vi løkken på en litt annen måte enn i DNA-eksemplet.
Der var neste verdi bestemt av et element i en eksisterende liste.
Her er neste verdi bestemt av den forrige verdien vi selv nettopp beregnet.

Det er en veldig viktig idé i modellering.

## Hvor realistisk er modellen?

Det er viktig å vite hva modellen *ikke* tar hensyn til.

* modellen antar at alle smittede smitter videre fra første dag
* modellen antar at ingen blir friske eller immune
* modellen antar at ingen tiltak blir satt inn
* modellen antar at den samme vekstfaktoren gjelder hele tiden

Likevel er modellen nyttig.
Den hjelper oss å forstå hva eksponentiell vekst betyr i praksis,
og hvorfor tiltak tidlig i et utbrudd kan være så viktige.
