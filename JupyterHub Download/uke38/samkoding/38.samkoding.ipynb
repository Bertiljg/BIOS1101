### La oss se på np.arange versus range()

np.arange er en numpy array construktør som gir en numpy array av tall med gitt antall elementer.
np.array brukes slik:

np.arange(**start**, **stopp**, **steg**)

veldig likt:

range(**start**, **stopp**, **steg**)

Begge konstruktørene gir start ved start verdien, men stopper før stopp verdien! Forskjellen mellom np.arange() og range() er at np.arange() kan ta flyttal (float) som input-argumenter, mens range() bare kan ta heltall (int).

## La oss se på et eksempel dere kjenner fra før,  Algevekst i vann fra uke 36.

En vekstmodell som representerer arealet, $A$ av innsjøen som er dekket av alger en gitt dag er gitt ved:

<!-- Equation labels as ordinary links -->
<div id="eq:modell"></div>

$$A_t = (1 + r) A_{t-1}$$

For algeveksten i innsjøen er $r=1$ når tidssteget er 1 dag.

I svar til oppgaven skal dere ha kommet frem til en implementering som ser ut som dette:

```
# Tidliger skrevet kode! Ikke fjern denne fra samkodings bolk.
import matplotlib.pyplot as plt

T = 30          # simulasjonstid (dager)
r = 1           # vekstrate (daglig)
A_maks = 10     # maks areal, km2
A0_m2 = 0.018   # startarel, m2
t0 = 1          # startdag, 1. juni

A = []        # liste for å lagre areal over tid
tid = []      # liste for å lagre tidspunkter

# konvertere startareal fra m2 til km2
A0_km2 = A0_m2 / 1e6
""" La oss se om vi kunne skrive denne delen anderledes med numpy """
# startbetingelser
A.append(A0_km2)
tid.append(t0)

# simulasjonsløkke
t = 1         # løkkevariabel

while t <= T:    
    A_ny = (1 + r) * A[t-1]
    A.append(A_ny)
    tid_ny = tid[t-1] + 1
    tid.append(tid_ny)
    t = t + 1

""" """
# plot formatering...

plt.plot(tid, A, label='Vekstkurve')
plt.axhline(A_maks, ls=':', color='k', label='Maks innsjøareal')
plt.axvline(tid[-2], ls=':', color='r', label='Dag innsjø er dekket')
plt.xlabel('Dag')
plt.ylabel('Areal dekket i km2')
plt.title('Algevekst i innsjø')
plt.legend()
plt.show()
```

Siden hvert nye element i kjeden er definert med egenskapen :

$$A_{ny}  = (1 + r) * A[t-1] $$

kan dette faktisk skrives om som at punktet $A_n$ har følgende egenskap:

$$A_n = (1 + r)^n A_0$$

$$A_n = A_0 *(1 + r)^n$$

Slik at en rekken av tall kan regnes ut så enkelt ved bruk av numpy!

## Det er mulig å bestemme a og b verdiene for en rett linje 

Funksjonen under viser hvordan vi kan bestemme a og b verdiene for en rett linje! 

Det første som er klart at en rett linje her må gå igjennom snittet av x verdiene og snittet av y verdiene!

<!-- begin figure -->

<img src="Gresshoppe7.png" width="600"><p style="font-size: 0.9em"><i>Figur 8: Alle punktene og avstand til linje sammen med senter punkt.</i></p>

Vi går ikke ytterliggere inn på matematikken bak dette her, utover å vise en funksjon:

$$\large a = \frac{\sum{(x_{i}} - \bar{x})*(y_{i} - \bar{y})}{\sum{(x_{i}} - \bar{x})^2} $$

and

$$\large b = \bar{y} - a  \bar{x} $$

hvor $\large \bar{x}$ er gjennomsnittlig x verdi og $\large \bar{y}$ er gjennomsnittlig y verdi.
